# Visualize Road Edges and Object Trajectories for Collision Checking

```python
import json
import trimesh
```

```python
f = open('../data/processed/training/tfrecord-00000-of-01000_0.json', 'r')
scene = json.load(f)
```

```python
scene.keys()
```

```python
# Meshes for collision checking
def generate_mesh(segments, radius=0.1, color=None):
    cylinders = []
    for segment in segments:
        start, end = segment
        cylinder = trimesh.creation.cylinder(
            radius=radius, 
            segment=[start, end]
        )
        if color is not None:
            cylinder.visual.face_colors = color

        cylinders.append(cylinder)

    mesh = trimesh.util.concatenate(cylinders)
    return mesh
```

```python
# Construct road edges
edge_segments = []
for road in scene['roads']:
    if road["type"] == 'road_edge':
        edge_vertices = [[r["x"], r["y"], r["z"]] for r in road["geometry"]]
        edge_segments += [[edge_vertices[i], edge_vertices[i+1]] for i in range(len(edge_vertices) - 1)]
edge_mesh = generate_mesh(edge_segments)
```

```python
# Construct objects
object_meshes = []
trajectory_segments_list = []
for object in scene['objects']:
    if False in object["valid"]:
        object_vertices = [[pos["x"], pos["y"], pos["z"]] for pos in object["position"][:object["valid"].index(False)]]
    else:
        object_vertices = [[pos["x"], pos["y"], pos["z"]] for pos in object["position"]]
    if len(object_vertices) > 1:
        trajectory_segments = [[object_vertices[i], object_vertices[i+1]] for i in range(len(object_vertices) - 1)]
        trajectory_segments_list.append(trajectory_segments)
        if object["mark_as_static"] == True:
            print(len(object["position"]))  
            trajectory_mesh = generate_mesh(trajectory_segments, color=(255,0,0,180))
        else:
            trajectory_mesh = generate_mesh(trajectory_segments, color=(0,255,0,180))
        object_meshes.append(trajectory_mesh)
```

```python
# Render
view = trimesh.Scene()
view.add_geometry(edge_mesh, node_name="Road Edges")
for i, mesh in enumerate(object_meshes):
        view.add_geometry(mesh, node_name=f"Vehicle {i} Path")
view.show()
```
